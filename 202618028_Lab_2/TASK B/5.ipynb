### Task 5 - Filtering and Querying

```python
import pandas as pd

data = pd.read_csv("train.csv")


male_older_50 = data[(data["Sex"] == "male") & (data["Age"] > 50)]

print("1. Male passengers older than 50:", len(male_older_50))


female_first_class = data[(data["Sex"] == "female") & (data["Pclass"] == 1)]

survival_percentage = female_first_class["Survived"].mean() * 100

print("2. Female first-class passengers:", len(female_first_class))
print("   Percentage survived:", survival_percentage)



fare_median = data["Fare"].median()

group3 = data[
    (data["Age"] >= 20) &
    (data["Age"] <= 40) &
    (data["Fare"] > fare_median) &
    (data["Survived"] == 1)
]

print("3. Age 20-40, fare above median, survived:", len(group3))



group4 = data[
    (data["SibSp"] == 0) &
    (data["Parch"] == 0) &
    (data["Age"] < 30) &
    (data["Survived"] == 0)
]

print("4. Travelling alone, age < 30, did not survive:", len(group4))



southampton = data[data["Embarked"] == "S"]

southampton_median = southampton["Fare"].median()

group5 = data[
    (data["Embarked"] == "S") &
    (data["Pclass"].isin([2, 3])) &
    (data["Fare"] > southampton_median)
]

print("5. Embarked S, Pclass 2/3, fare above Southampton median:", len(group5))
```
