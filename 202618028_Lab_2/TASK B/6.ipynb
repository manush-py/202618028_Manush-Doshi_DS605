### Task 6 - groupby() and Aggregation

```python
import pandas as pd

data = pd.read_csv("train.csv")


survival_by_sex = data.groupby("Sex")["Survived"].mean() * 100

print("1. Survival Rate by Sex:")
print(survival_by_sex)


survival_by_class = data.groupby("Pclass")["Survived"].mean() * 100

print("\n2. Survival Rate by Pclass:")
print(survival_by_class)


average_by_class = data.groupby("Pclass")[["Age", "Fare"]].mean()

print("\n3. Average Age and Fare by Pclass:")
print(average_by_class)


sex_class = data.groupby(["Sex", "Pclass"]).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)


sex_class["Survival_Rate"] = sex_class["Survival_Rate"] * 100

print("\n4. Passenger Count and Survival Rate by Sex-Pclass:")
print(sex_class)


embarked_stats = data.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)


embarked_stats["Survival_Rate"] = embarked_stats["Survival_Rate"] * 100

print("\n5. Passenger Count, Average Fare, and Survival Rate by Embarked:")
print(embarked_stats)
```
